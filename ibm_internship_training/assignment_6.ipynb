## Task 7: Diagnose Data Leakage

Problems:

1. Scaling before train-test split.

   * Leakage because information from test data influences scaling.

2. amount_paid_after_purchase used to predict will_purchase.

   * Leakage because this information is only known after purchase.

3. No train-test split before preprocessing.

   * Test information contaminates training.

Correct Workflow:

1. Remove leakage columns.
2. Separate features and target.
3. Train-test split.
4. Fit preprocessing only on training data.
5. Transform training data.
6. Transform test data using the same fitted preprocessing.
7. Train model on training data.
8. Evaluate on test data.
