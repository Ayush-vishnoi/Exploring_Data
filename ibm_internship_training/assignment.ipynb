```python
image = np.array([
    [255, 128,  64,  32],
    [ 10, 200, 150,  80],
    [ 45, 170,  90, 220],
    [100,  55, 185, 240]
])
flat = image.flatten()             
normalised = flat / 255            

print("Flattened length:", len(flat))
print("Max after normalise:", normalised.max())
print("Normalised:", normalised.round(2))
```

```python
sales = np.array([
    [200, 150, 300, 180],
    [220, 190, 280, 210],
    [250, 170, 320, 240]
])
print("Total per product:", sales.sum(axis=0))
print("Total per month  :", sales.sum(axis=1))
totals = sales.sum(axis=1)
best = max(totals)
print("Best selling month total: ₹", best)
```
