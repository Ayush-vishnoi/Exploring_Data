```python
df.head()
```

```python
dept=df.groupby("Department")['Salary_INR'].mean().astype(int)
```

```python
dept.plot(x=dept.index,y=dept.values,color='c',kind='barh')
plt.grid()
plt.title('Aerage salary by departmnet')
plt.ylabel("Salary")
plt.show()
```

```python
salary=df.groupby("Job_Title")['Salary_INR'].mean().astype(int).sort_values().tail(10)
plt.figure(figsize=(15,4))
salary.plot(x="salary.index",y="salary.values",kind="bar",color="m")
plt.title('Aerage salary by job_title')
plt.ylabel("Salary")
plt.show()
```

```python
df[df.Status.isin(["Resigned","Terminated"])].value_counts()
```

```python
df[df.Status.isin(["Resigned","Terminated"])].groupby("Job_Title")["Status"].value_counts().unstack().plot(kind="bar", color=["red","orange"], figsize=(15,6))
plt.title("Resigned vs Terminated by Job Title")
plt.xlabel("Job Title")
plt.ylabel("Count")
plt.xticks(rotation=45, ha="right")
plt.legend(title="Status")
plt.show()
```
