# Q1.Flatten a nested list

```python
nested = [[1, 2], [3, 4], [5, 6]]

flat = []

for sublist in nested:
    for num in sublist:
        flat.append(num)

print(flat)
```

# Q2.Word frequency counter

```python
sentence = input("Enter a sentence: ")
print(sentence)

words = sentence.lower().split()
freq = {}

for word in words:
    freq[word] = freq.get(word, 0) + 1

print(freq)
```

# Q3. calculator with function

```python
def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

def multiply(a, b):
    return a * b

def divide(a, b):
    if b == 0:
        return "Cannot divide by zero"
    return a / b

while True:
    print("\n1. Add")
    print("2. Subtract")
    print("3. Multiply")
    print("4. Divide")
    print("5. Exit")

    choice = int(input("Enter choice: "))

    if choice == 5:
        break

    a = float(input("Enter first number: "))
    b = float(input("Enter second number: "))

    if choice == 1:
        print("Result:", add(a, b))
    elif choice == 2:
        print("Result:", subtract(a, b))
    elif choice == 3:
        print("Result:", multiply(a, b))
    elif choice == 4:
        print("Result:", divide(a, b))
    else:
        print("Invalid choice")
```

# Q4.Marks Anaylyzer

```python
def analyze_marks(marks):
    minimum = min(marks)
    maximum = max(marks)
    average = sum(marks) / len(marks)

    passed = len([m for m in marks if m >= 40])
    pass_percentage = (passed / len(marks)) * 100

    grades = []
    for m in marks:
        if m >= 90:
            grades.append("A")
        elif m >= 75:
            grades.append("B")
        elif m >= 60:
            grades.append("C")
        elif m >= 40:
            grades.append("D")
        else:
            grades.append("F")

    return minimum, maximum, average, pass_percentage, grades

marks = [85, 92, 67, 45, 30]

result = analyze_marks(marks)

print("Min:", result[0])
print("Max:", result[1])
print("Average:", result[2])
print("Pass %:", result[3])
print("Grades:", result[4])
```
