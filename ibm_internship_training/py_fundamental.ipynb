# Q3. calculator with function

```python
def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

def multiply(a, b):
    return a * b

def divide(a, b):
    if b == 0:
        return "Cannot divide by zero"
    return a / b

while True:
    print("\n1. Add")
    print("2. Subtract")
    print("3. Multiply")
    print("4. Divide")
    print("5. Exit")

    choice = int(input("Enter choice: "))

    if choice == 5:
        break

    a = float(input("Enter first number: "))
    b = float(input("Enter second number: "))

    if choice == 1:
        print("Result:", add(a, b))
    elif choice == 2:
        print("Result:", subtract(a, b))
    elif choice == 3:
        print("Result:", multiply(a, b))
    elif choice == 4:
        print("Result:", divide(a, b))
    else:
        print("Invalid choice")
```

# Q4.Marks Anaylyzer

```python
def analyze_marks(marks):
    minimum = min(marks)
    maximum = max(marks)
    average = sum(marks) / len(marks)

    passed = len([m for m in marks if m >= 40])
    pass_percentage = (passed / len(marks)) * 100

    grades = []
    for m in marks:
        if m >= 90:
            grades.append("A")
        elif m >= 75:
            grades.append("B")
        elif m >= 60:
            grades.append("C")
        elif m >= 40:
            grades.append("D")
        else:
            grades.append("F")

    return minimum, maximum, average, pass_percentage, grades

marks = [85, 92, 67, 45, 30]

result = analyze_marks(marks)

print("Min:", result[0])
print("Max:", result[1])
print("Average:", result[2])
print("Pass %:", result[3])
print("Grades:", result[4])
```

```python
import random
rand=[]

for i in range(10):
    x=random.uniform(50,100)
    rand.append(x)
print(rand)
sumi=0
maxi=rand[0]
mini=rand[0]
for num in rand:
    if num > maxi:
        maxi = num
    if num < mini:
        mini = num
    sumi += num
print("Max:",maxi)
print("Min:",mini)
print("Mean:",sumi/10.0)
```

# Q4 oop challenge

```python
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance
    def deposit(self, amount):
            self.balance += amount
            print(f"Deposited ₹{amount} -> Balance: ₹{self.balance}")
    def withdraw(self, amount):
        try:
            if amount > self.balance:
                raise Exception("Overdraft")
            self.balance -= amount
            print("Withdrawn:", amount)
        except Exception as e:
            print("Error:", e)
    def get_balance(self):
        print(f"Balance: ₹{self.balance}")
        
acc = BankAccount("Ayush", 1000)
acc.deposit(500)
acc.withdraw(200)
acc.get_balance()
acc.withdraw(5000)
```

```python
employee = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve', 'Frank', 'Grace', 'Henry', 'Isla', 'Jack'],
    'Age': [30, 25, 35, 28, 40, 32, 27, 45, 29, 38],
    'Salary': [70000, None, 90000, 60000, 110000, None, 55000, None, 65000, 95000],
    'Experience': [6, 3, None, 4, 12, 7, 2, 10, None, 8]
})
employee.to_csv('employees.csv', index=False)
df = pd.read_csv('employees.csv')
print("Original:\n", df.head(10))
```

```python
df['Salary'] = df['Salary'].fillna(df['Salary'].mean())
df['Experience'] = df['Experience'].fillna(0)
```

```python
df.head()
```

```python
seniority = []
for x in df['Experience']:
    if x >= 5:
        seniority.append('Senior')
    else:
        seniority.append('Junior')

df['Seniority'] = seniority
```

```python
print("\nCleaned Data:\n", df)
print("\nAverage Salary:\n", df.groupby('Seniority')['Salary'].mean().round(2))
```
